```python
import numpy as np
import matplotlib.pyplot as plt
import os

print("Notebook is working.")
print("Current directory:", os.getcwd())
```

```python
from pathlib import Path

runs_dir = Path("../runs")
all_npz = list(runs_dir.glob("run_*/traj.npz"))

print(f"Found {len(all_npz)} traj files")

for p in all_npz[:5]:
    print(p)
```

```python
import numpy as np

if len(all_npz) == 0:
    print("No traj files found.")
else:
    data = np.load(all_npz[0])
    print(data.files)
```

```python
r = data["r"]
vr = data["vr"]
target_r = float(data["target_r"])

print("len(r) =", len(r))
print("len(vr) =", len(vr))
print("target_r =", target_r)
print("final r =", r[-1])
print("final vr =", vr[-1])
print("final radius error =", r[-1] - target_r)
print("avg abs radius error =", np.mean(np.abs(r - target_r)))
```

```python
plt.figure(figsize=(10, 4))
plt.plot(r)
plt.axhline(target_r, linestyle="--")
plt.title("Radius vs Time")
plt.xlabel("Step")
plt.ylabel("Radius")
plt.show()

plt.figure(figsize=(10, 4))
plt.plot(vr)
plt.axhline(0, linestyle="--")
plt.title("Radial Velocity vs Time")
plt.xlabel("Step")
plt.ylabel("Radial Velocity")
plt.show()
```
